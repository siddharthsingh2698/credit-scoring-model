# Phase 4: Exploratory Data Analysis (EDA)

This notebook visualizes the German Credit dataset to understand the feature distributions and relationships with credit risk.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

# Set plotting style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Load data
df = pd.read_csv('../data/raw/german_credit_dataset.csv')
df.head()
```

## 1. Target Distribution
Let's look at the balance of our target variable `class`.

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='class', palette='viridis')
plt.title('Target Variable Distribution (class)')
plt.show()

print(df['class'].value_counts(normalize=True) * 100)
```

## 2. Numerical Features Distribution
Visualizing the distributions of numerical features like `credit_amount`, `duration`, and `age`.

```python
num_cols = ['credit_amount', 'duration', 'age']

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for i, col in enumerate(num_cols):
    sns.histplot(df[col], kde=True, ax=axes[i], color='skyblue')
    axes[i].set_title(f'Distribution of {col}')
plt.tight_layout()
plt.show()
```

## 3. Relationships with Creditworthiness (Target)
How do numerical features differ between 'good' and 'bad' credit risks?

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for i, col in enumerate(num_cols):
    sns.boxplot(data=df, x='class', y=col, palette='Set2', ax=axes[i])
    axes[i].set_title(f'{col} by class')
plt.tight_layout()
plt.show()
```

## 4. Categorical Features
Let's inspect some important categorical variables like `checking_status` and `credit_history`.

```python
cat_cols = ['checking_status', 'credit_history', 'purpose']

for col in cat_cols:
    plt.figure(figsize=(10, 5))
    sns.countplot(data=df, y=col, hue='class', palette='Set1')
    plt.title(f'{col} by class')
    plt.tight_layout()
    plt.show()
```

## 5. Correlation Heatmap
Checking for correlations between numerical variables.

```python
numeric_df = df.select_dtypes(include=[np.number])
plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
## 6. Missing Values and Risk Patterns

The final part of Phase 4 focuses on data quality and on the relationship between applicant attributes and credit risk.

We will check whether any values are missing and compare the most important categorical and numeric variables across the two classes.
```

```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns

# Missing values check
missing_values = df.isna().sum()
print('Missing values by column:')
print(missing_values[missing_values > 0] if (missing_values > 0).any() else 'No missing values detected.')

# Numeric features most associated with risk
num_features = ['duration', 'credit_amount', 'age', 'installment_commitment', 'residence_since', 'existing_credits']

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()
for ax, col in zip(axes, num_features):
    sns.boxplot(data=df, x='class', y=col, palette='Set2', ax=ax)
    ax.set_title(f'{col} by class')
    ax.set_xlabel('Class')
    ax.set_ylabel(col)
plt.tight_layout()
plt.show()

# Categorical patterns by class
cat_features = ['checking_status', 'credit_history', 'savings_status', 'housing']
for col in cat_features:
    plt.figure(figsize=(10, 5))
    sns.countplot(data=df, x=col, hue='class', palette='Set1')
    plt.title(f'{col} by class')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

# Scatterplots for risk-related relationships
scatter_pairs = [('duration', 'credit_amount'), ('age', 'credit_amount'), ('duration', 'installment_commitment')]
for x_col, y_col in scatter_pairs:
    plt.figure(figsize=(7, 5))
    sns.scatterplot(data=df, x=x_col, y=y_col, hue='class', palette='Set2', alpha=0.7)
    plt.title(f'{x_col} vs {y_col} by class')
    plt.tight_layout()
    plt.show()

# Summary of key findings
print('\nKey Phase 4 findings:')
print('- There are no missing values in the dataset, so the quality check passes.')
print('- Higher credit amounts and longer durations are more common in the bad-risk group.')
print('- Applicants with risky checking/savings profiles and poor credit histories show higher default risk.')
print('- The relationship between credit behavior and class is stronger than simple age patterns alone.')
```
